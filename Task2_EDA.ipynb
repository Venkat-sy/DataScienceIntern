# Task 2: Exploratory Data Analysis (EDA)

**Objective:** Perform EDA on a sample dataset using pandas and matplotlib/seaborn. Create visualizations (histograms, bar charts, scatter plots) and identify patterns/trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD DATASET
try:
    df = pd.read_csv("cleaned_employee_data.csv")
    print("--- Dataset Loaded Successfully ---")
    print(df.head(), "\n")
    
    # Configure plot style
    sns.set_theme(style="whitegrid")
    
    # 2. CREATE VISUALIZATIONS
    fig, axs = plt.subplots(2, 2, figsize=(16, 12))
    
    # Histogram: Age
    sns.histplot(df["Age"], bins=5, kde=True, color="skyblue", ax=axs[0, 0])
    axs[0, 0].set_title("Distribution of Employee Ages")
    
    # Bar Chart: Department Count
    sns.countplot(data=df, x="Department", palette="viridis", ax=axs[0, 1])
    axs[0, 1].set_title("Number of Employees per Department")
    
    # Bar Chart: Avg Salary
    sns.barplot(data=df, x="Department", y="Salary", errorbar=None, palette="magma", ax=axs[1, 0])
    axs[1, 0].set_title("Average Salary by Department")
    
    # Scatter Plot: Age vs Salary
    sns.scatterplot(data=df, x="Age", y="Salary", hue="Department", s=150, ax=axs[1, 1])
    axs[1, 1].set_title("Age vs. Salary")
    
    plt.tight_layout()
    plt.show()
    
except FileNotFoundError:
    print("Error: 'cleaned_employee_data.csv' not found. Please run Task 1 first!")
```

### EDA Key Findings Writeup

1. **Age Distribution:** The histogram shows that the majority of our workforce is heavily concentrated between the ages of 25 and 32.
2. **Department Demographics:** The IT department is the largest team in the company, followed closely by Sales and HR.
3. **Salary Disparities:** The bar chart reveals that the HR and Sales departments have the highest average compensation.
4. **Age vs. Salary Trend:** The scatter plot illustrates a generally positive correlation between Age and Salary—older employees tend to earn higher salaries.
