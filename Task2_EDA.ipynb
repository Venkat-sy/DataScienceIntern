# Task 2: Exploratory Data Analysis (EDA)

**Objective:** Perform EDA on a sample dataset using pandas and matplotlib/seaborn. Create visualizations (histograms, bar charts, scatter plots) and identify patterns/trends.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the cleaned dataset from Task 1
df = pd.read_csv("cleaned_employee_data.csv")
display(df.head())
```

### 1. Histogram: Age Distribution

```
plt.figure(figsize=(8, 5))
sns.histplot(df["Age"], bins=5, kde=True, color="skyblue")
plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### 2. Bar Chart: Employee Count by Department

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Department", palette="viridis")
plt.title("Number of Employees per Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

### 3. Bar Chart: Average Salary by Department

```
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Department", y="Salary", ci=None, palette="magma")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary ($)")
plt.show()
```

### 4. Scatter Plot: Age vs. Salary

```
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Salary", hue="Department", s=100)
plt.title("Age vs. Salary")
plt.xlabel("Age")
plt.ylabel("Salary ($)")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

### EDA Key Findings Writeup

Based on the Exploratory Data Analysis (EDA) of the employee dataset, we observed the following key patterns and trends:

1. **Age Distribution:** The histogram shows that the majority of our workforce is relatively young, heavily concentrated between the ages of 25 and 32. There is a slight right skew, with fewer older employees (above 35).
2. **Department Demographics:** The IT department is the largest team in the company, followed closely by Sales and HR. Marketing currently has the fewest employees in this sample.
3. **Salary Disparities:** The bar chart analyzing average salary by department reveals that the HR and Sales departments have the highest average compensation. The Marketing department shows the lowest average salary.
4. **Age vs. Salary Trend:** The scatter plot illustrates a generally positive correlation between Age and Salary—older employees tend to earn higher salaries, which is consistent with accumulating more experience over time. However, it also highlights potential outliers, such as a younger employee in HR earning a top-tier salary compared to their peers.
