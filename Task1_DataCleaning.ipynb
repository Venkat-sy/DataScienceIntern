# Task 1: Data Cleaning & Exploration with Pandas

**Objective:** Load a small sample dataset, handle missing values, duplicates, and incorrect data types. Display basic statistics and summarize the cleaning steps.

```
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv("sample_employee_data.csv")
display(df.head())
```

### Initial Exploration

```
# Check info for data types and null values
df.info()
```

```
# Display basic statistics before cleaning
display(df.describe())
```

### Data Cleaning

```
# 1. Handle Duplicates
print(f"Duplicates before: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Duplicates after: {df.duplicated().sum()}")
```

```
# 2. Handle Missing Values
print("\nMissing values before:\n", df.isnull().sum())

# Fill missing Age with median age
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)

# Fill missing Salary with the mean salary of their department
df["Salary"] = df.groupby("Department")["Salary"].transform(lambda x: x.fillna(x.mean()))

print("\nMissing values after:\n", df.isnull().sum())
```

```
# 3. Fix Data Types
# Convert Join_Date from string to datetime object
df["Join_Date"] = pd.to_datetime(df["Join_Date"])
df.info()
```

### Final Statistics & Value Counts

```
# Basic statistics of the cleaned data
display(df.describe())

# Value counts for categorical column
print("\nDepartment Value Counts:")
print(df["Department"].value_counts())
```

```
# Save cleaned dataset for Task 2
df.to_csv("cleaned_employee_data.csv", index=False)
```

### Data Cleaning Summary Report

**1. Loading & Inspection:** We loaded `sample_employee_data.csv` using pandas. The initial inspection via `.info()` revealed missing values in the `Age` and `Salary` columns, duplicated rows, and an incorrect data type for `Join_Date` (stored as an object/string instead of a datetime).

**2. Removing Duplicates:** We identified and removed 2 duplicate rows using `drop_duplicates()`, ensuring each employee is only represented once.

**3. Handling Missing Values:** 
- **Age:** Missing age values were imputed using the overall **median** age to avoid the influence of outliers.
- **Salary:** Missing salaries were imputed using the **mean** salary of the respective employee's department to provide a more accurate and contextual estimation.

**4. Fixing Data Types:** The `Join_Date` column was successfully cast to a pandas `datetime` object, which will allow for time-based analysis in the future.

**5. Results:** The final dataset has 0 missing values, 0 duplicates, and correct data types across all columns. Basic statistics confirm the distributions are now clean and ready for Exploratory Data Analysis (EDA).
