# Task 1: Data Cleaning & Exploration with Pandas

**Objective:** Load a small sample dataset, handle missing values, duplicates, and incorrect data types. Display basic statistics and summarize the cleaning steps.

```python
import pandas as pd
import numpy as np

# 1. CREATE AND LOAD DATASET
data_1 = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 103, 108],
    "Age": [25, np.nan, 28, 35, 42, 29, np.nan, 31, 26, 38, 28, 31],
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT", "Sales", "Marketing", "Marketing", "Sales", "IT", "Marketing"],
    "Salary": [60000, 55000, 62000, 75000, 80000, np.nan, 71000, 68000, 65000, 72000, 62000, 68000],
    "Join_Date": ["2020-01-15", "2019-03-22", "2021-07-01", "2018-11-12", "2015-05-30", "2022-02-18", "2020-09-09", "2019-12-01", "2021-04-14", "2017-08-25", "2021-07-01", "2019-12-01"]
}
df = pd.DataFrame(data_1)
df.to_csv("sample_employee_data.csv", index=False)

print("--- 1. INITIAL DATASET ---")
print(df.head(), "\n")
print(f"Duplicates: {df.duplicated().sum()}")
print("Missing Values:\n", df.isnull().sum(), "\n")

# 2. DATA CLEANING
# Drop duplicates
df = df.drop_duplicates()

# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with Department mean
df["Salary"] = df.groupby("Department")["Salary"].transform(lambda x: x.fillna(x.mean()))

# Fix Data Type
df["Join_Date"] = pd.to_datetime(df["Join_Date"])

# 3. RESULTS
print("--- 2. CLEANED DATASET ---")
print(f"Duplicates: {df.duplicated().sum()}")
print("Missing Values:\n", df.isnull().sum(), "\n")
print("--- Basic Statistics ---")
print(df.describe())

# Save for Task 2
df.to_csv("cleaned_employee_data.csv", index=False)
print("\nSUCCESS: Cleaned data saved to 'cleaned_employee_data.csv'")
```

### Data Cleaning Summary Report

**1. Loading & Inspection:** We generated and loaded `sample_employee_data.csv`. Initial inspection revealed missing values in `Age` and `Salary`, duplicated rows, and an incorrect data type for `Join_Date`.

**2. Removing Duplicates:** We removed duplicate rows, ensuring each employee is only represented once.

**3. Handling Missing Values:** 
- Missing age values were imputed using the overall **median** age.
- Missing salaries were imputed using the **mean** salary of the respective department.

**4. Fixing Data Types:** `Join_Date` was cast to a datetime object.

**5. Results:** The final dataset has 0 missing values, 0 duplicates, and correct data types.
